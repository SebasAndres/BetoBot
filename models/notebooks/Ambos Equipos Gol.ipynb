## Probabilidad ambos equipos hacen gol.

<image src="diagrams/AmbosGol.png" height=500>

Requerimientos:
-  Tener un registro de la cantidad de goles por partido por equipo

    $X_E$ := [$partido_1, ... partido_T$]

- Tener un registro de de la cantidad de goles recibidos por partido por equipo

    $Q_E$ := [$partido_1, ... partido_T$]

- Tener un registro de la cantidad de partidos con goles de ambos equipos por liga.

    $Z_L$ := [$partido_1, ... partido_T$]


### Modelos

> $X_{E}$ ~ $Be(p)$: El equipo E hace al menos un gol en un partido. \
> $\bar X_{E}$ ~ $Bi(n,p)$: Cantidad de partidos en los que E hizo al menos un gol en N partidos. \
> $Q_{E}$ ~ $Be(p)$: El equipo E no recibe un gol en el partido.   
> $\bar Q_{E}$ ~ $Bi(n, p)$: Cantidad de partidos en los que E no recibio un gol. \
> $Z_L$ ~ $Bi(n,p_z)$: Cantidad de partidos en los que hay un gol de los dos equipos en la liga L.


$f$ modela la probabilidad de que sea buena la apuesta:

> $f:[0,1]^7\rightarrow [0,1]$ \
> $f(X_{E_1}, X_{E_2}, Q_{E_1}, Q_{E_2}, P(X_{E_1}\cap X_{E_2}), P(Q_{E_1}\cap Q_{E_2}), p_z) \in [0,1]$ 

Donde: 

1. Probabilidad conjunta de que hagan gol: $P(X_{E_1}\cap X_{E_2}) = P(X_{E_i}|X_{E_j})*P(X_{E_j})$ 

2. Probabilidad de que no haya goles en el partido: $P(Q_{E_1}\cap Q_{E_2}) = P(Q_{E_i}|Q_{E_j})*P(Q_{E_j})$

3. Probabilidad de que ambos equipos cualquiera hagan gol en la liga.

### Entrenamiento del modelo

Estimacion de $p_z$ por EMV:
$p_z = \frac{\sum_i^N{Z_i}}{N^2}$

Para obtener $P(K_{E_1}\cap K_{E_2})$, vamos a utilizar probabilidades condicionales $P(K_i|K_j)$ calculadas en base a los datos.
