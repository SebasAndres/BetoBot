## Total X por partido.

Requerimientos:
- Tener un registro de la cantidad de amarillas de un equipo por partido en una liga. 

  $X_E$ := [$partido_1,...,partido_T$]
- Tener un registro de cantidad de amarillas por partido (de cada lado) en la liga.

  $Y_L$ := $[(EQ_1,EQ_2), (EQ_3,EQ_8), ...]$


### Modelo

Suponemos $X_E$ ~ $Poisson(\lambda)$,

$P(X_1+X_2 > U) = \sum_{k=0}^U P(X_1=k,X_2=U-k)$

#### (v1) independencia:
$P(X_1+X_2 > U) = \sum_{k=0}^U P(X_1=k)P(X_2=U-k) = ... = (\frac{\lambda_2^U*e^{-\lambda_1-\lambda_2}}{U!})\sum_{k=0}^{U}(\frac{\lambda_1}{\lambda_2})^k \binom{U}{k}$

#### (v2) sin independencia:
Tablas por liga de probabilidad conjunta, aproximadas por:
$P(X_1=A,X_2=B) = $ {# $(A,B) \lor (B,A)$} / $N$
